## Describe 3 methods you can do with trees in biopython

The collapse method allows you to delete a target and relink the children back to the parent of the target.

The ladderize method allows you to sort clades by the number of terminal nodes it has. By default the deepest clades are placed last.

The split method allows the user to take a particular clade and split it into subclades.
